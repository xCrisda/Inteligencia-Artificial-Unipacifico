# Creación de un DataFrame

## ¿Qué es un DataFrame?

Un DataFrame es una estructura de datos bidimensional organizada en filas y columnas.

Su funcionamiento es similar al de una tabla de Excel o una base de datos, permitiendo almacenar y manipular información de manera estructurada.

En este ejercicio se creará un DataFrame con información de series y películas de una plataforma de streaming.

````python
# ======================================================
# CREACIÓN DE UN DATAFRAME
# ======================================================

# Información de películas y series
streaming = {
    "Título": [
        "Dark",
        "Breaking Bad",
        "Interstellar",
        "Stranger Things",
        "Inception"
    ],

    "Tipo": [
        "Serie",
        "Serie",
        "Película",
        "Serie",
        "Película"
    ],

    "Año": [2017, 2008, 2014, 2016, 2010],

    "Calificación": [9.0, 9.5, 8.7, 8.8, 8.8]
}

# Crear DataFrame
df_streaming = pd.DataFrame(streaming)

# Mostrar DataFrame
df_streaming
````

## Análisis del resultado

El DataFrame organiza información relacionada con películas y series.

Las columnas utilizadas fueron:

- Título
- Tipo
- Año
- Calificación

Este tipo de estructuras es utilizado por plataformas digitales para almacenar y analizar información de usuarios y contenido multimedia.

# Operaciones entre columnas

Pandas permite realizar operaciones matemáticas y crear nuevas columnas automáticamente.

En este ejercicio se realizará:

- Conversión de calificaciones a porcentaje.
- Clasificación del contenido según su puntuación.

````python
# ======================================================
# OPERACIONES ENTRE COLUMNAS
# ======================================================

# Convertir calificación a porcentaje
df_streaming["Porcentaje"] = (
    df_streaming["Calificación"] * 10
)

# Clasificar contenido
df_streaming["Popularidad"] = np.where(
    df_streaming["Calificación"] >= 9.0,
    "Muy popular",
    "Popular"
)

# Mostrar resultados
df_streaming
````

## Análisis del resultado

Se realizaron operaciones utilizando la columna de calificación.

Posteriormente se crearon nuevas columnas:

- Porcentaje
- Popularidad

Esto permite analizar automáticamente qué contenido tiene mejores valoraciones dentro de la plataforma.

# Lectura y análisis de archivos CSV

## ¿Qué es un archivo CSV?

Un archivo CSV (Comma Separated Values) es un formato utilizado para almacenar datos en forma de tabla utilizando texto plano.

Cada fila representa un registro y cada columna representa una característica específica.

Los archivos CSV son ampliamente utilizados en:

- Ciencia de datos
- Inteligencia artificial
- Bases de datos
- Sistemas empresariales
- Plataformas digitales

---

## Importancia de los archivos CSV

Los archivos CSV permiten almacenar grandes cantidades de información de forma ligera y compatible con múltiples programas como:

- Excel
- Python
- MySQL
- Power BI

Pandas facilita la lectura y análisis de estos archivos mediante funciones como:

```python
pd.read_csv()
```

Esto permite trabajar con información real proveniente de aplicaciones, empresas o plataformas digitales.

````python
# ======================================================
# CREACIÓN Y LECTURA DE ARCHIVOS CSV
# ======================================================

# Información de usuarios de streaming
usuarios_streaming = {

    "Usuario": [
        "Carlos",
        "Laura",
        "Miguel",
        "Sofía",
        "Andrés"
    ],

    "Horas_vistas": [
        120,
        95,
        140,
        80,
        110
    ],

    "Plan": [
        "Premium",
        "Básico",
        "Premium",
        "Estándar",
        "Premium"
    ],

    "Pago_mensual": [
        45,
        20,
        45,
        30,
        45
    ]
}

# Crear DataFrame
df_usuarios = pd.DataFrame(usuarios_streaming)

# Guardar archivo CSV
df_usuarios.to_csv(
    "usuarios_streaming.csv",
    index=False
)

print("Archivo CSV creado correctamente.")

# Leer archivo CSV
leer_csv = pd.read_csv(
    "usuarios_streaming.csv"
)

# Mostrar resultados
leer_csv
````

## Análisis del resultado

En este ejercicio se creó un archivo CSV con información relacionada con usuarios de una plataforma de streaming.

Posteriormente:

- La información fue almacenada en un archivo externo.
- El archivo fue leído nuevamente utilizando Pandas.
- Los datos fueron organizados automáticamente en un DataFrame.

Este procedimiento es ampliamente utilizado en ciencia de datos y análisis empresarial.

# Trabajo con datos sintéticos

## ¿Qué son los datos sintéticos?

Los datos sintéticos son datos generados artificialmente mediante programación.

Estos datos son utilizados cuando:

- No existen datos reales disponibles.
- Se desea realizar simulaciones.
- Se necesitan pruebas para modelos de inteligencia artificial.
- Se quiere evitar el uso de datos sensibles o privados.

En este ejercicio se simularán reproducciones y horas vistas dentro de una plataforma de streaming.

````python
# ======================================================
# GENERACIÓN DE DATOS SINTÉTICOS
# ======================================================

# Generar datos aleatorios
datos_streaming = {

    "Reproducciones": np.random.randint(
        1000,
        10000,
        12
    ),

    "Horas_vistas": np.random.randint(
        500,
        5000,
        12
    )
}

# Crear DataFrame
df_datos = pd.DataFrame(
    datos_streaming
)

# Agregar meses
df_datos["Mes"] = [

    "Enero",
    "Febrero",
    "Marzo",
    "Abril",
    "Mayo",
    "Junio",
    "Julio",
    "Agosto",
    "Septiembre",
    "Octubre",
    "Noviembre",
    "Diciembre"
]

# Calcular promedio de visualización
df_datos["Promedio_visualización"] = (
    df_datos["Horas_vistas"]
    /
    df_datos["Reproducciones"]
)

# Organizar columnas
df_datos = df_datos[

    [
        "Mes",
        "Reproducciones",
        "Horas_vistas",
        "Promedio_visualización"
    ]
]

# Mostrar resultados
df_datos
````

## Análisis del resultado

Se generaron datos aleatorios para simular el comportamiento de usuarios dentro de una plataforma de streaming.

Posteriormente se calcularon:

- Reproducciones mensuales.
- Horas vistas.
- Promedio de visualización.

Este tipo de análisis es utilizado en plataformas digitales para comprender el comportamiento de los usuarios y mejorar sistemas de recomendación basados en inteligencia artificial.

# Análisis estadístico

Pandas incluye herramientas estadísticas que permiten analizar información de manera rápida y eficiente.

La función:

```python
describe()

````python
# Estadísticas generales
estadisticas = df_datos.describe().round(2)

estadisticas
````

## Significado de las estadísticas obtenidas

La función `describe()` genera diferentes estadísticas importantes:

- **count:** cantidad de registros.
- **mean:** promedio de los datos.
- **std:** desviación estándar.
- **min:** valor mínimo.
- **25%:** primer cuartil.
- **50%:** mediana de los datos.
- **75%:** tercer cuartil.
- **max:** valor máximo.

Estas métricas permiten comprender mejor el comportamiento y distribución de la información analizada.

# Filtrado de datos

Pandas permite filtrar información específica dentro de un DataFrame.

Esto resulta útil para analizar únicamente los registros que cumplen ciertas condiciones.

````python
# ======================================================
# FILTRADO DE DATOS
# ======================================================

# Filtrar contenido con más de 5000 reproducciones
filtro = df_datos[
    df_datos["Reproducciones"] > 5000
]

# Mostrar resultados
filtro
````

## Análisis del resultado

En este ejercicio se filtraron únicamente los meses que tuvieron más de 5000 reproducciones.

El filtrado de datos permite analizar información específica y encontrar patrones relevantes dentro de grandes conjuntos de datos.
